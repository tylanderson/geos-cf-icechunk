```python
from query_thredds import recurse_catalog
from siphon.catalog import TDSCatalog

cat = TDSCatalog(
    "https://ds.nccs.nasa.gov/thredds/catalog/GMAO/GEOS-CF/analysis-v2/catalog.xml"
)
items = recurse_catalog(
    cat, pattern="Y*/M*/D*/[!c]*ana.aqc_tavg_1hr_glo_L1440x721*.nc4"
)
# list(islice(items, 2))
all_urls = list(items)
```

```python
len(all_urls)
```

```python
with open("geos-cf_v2_ana_urls.txt", "w") as f:
    for url in all_urls:
        f.write(url + "\n")
```
